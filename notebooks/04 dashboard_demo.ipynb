## 3. Test Data Availability

```python
print("\n📁 Checking Data Files Directly...")

# Check what data files actually exist
data_dirs = ['../data/interim', '../data/processed', '../data/raw']

for data_dir in data_dirs:
    path = Path(data_dir)
    if path.exists():
        print(f"✅ {data_dir}: EXISTS")
        files = list(path.glob('*'))
        for file in files:
            print(f"   - {file.name} ({file.stat().st_size} bytes)")
    else:
        print(f"❌ {data_dir}: DOES NOT EXIST")

# Try to load data files directly
try:
    interim_files = list(Path('../data/interim').glob('*.csv'))
    if interim_files:
        test_df = pd.read_csv(interim_files[0])
        print(f"✅ Direct CSV load: {len(test_df)} records from {interim_files[0].name}")
    else:
        print("❌ No CSV files in interim directory")
        
    processed_files = list(Path('../data/processed').glob('*.parquet'))
    if processed_files:
        test_skills = pd.read_parquet(processed_files[0])
        print(f"✅ Direct Parquet load: {len(test_skills)} records from {processed_files[0].name}")
    else:
        print("❌ No Parquet files in processed directory")
        
except Exception as e:
    print(f"❌ Direct file loading failed: {e}")
```

## 4. Test Analysis Functionality with Available Data
### 4.1 Initialize Analyzers with Real Data

```python
print("\n🎯 Testing Analyzers with Real Data...")

try:
    # Re-initialize with successfully loaded data
    data_processor = DataProcessor()
    df = data_processor.load_cleaned_data()
    skills_df = data_processor.create_skill_pivot(df)
    
    print(f"📊 Data ready: {len(df)} main records, {len(skills_df)} skill records")
    
    # Initialize all analyzers
    skill_analyzer = SkillAnalyzer()
    trend_analyzer = TrendAnalyzer() 
    role_analyzer = RoleAnalyzer()
    seniority_analyzer = SeniorityAnalyzer()
    ecosystem_analyzer = EcosystemAnalyzer()
    
    print("✅ All analyzers initialized with real data!")
    
except Exception as e:
    print(f"❌ Analyzer initialization failed: {e}")
```

### 4.2 Test Skill Analysis

```python
print("\n🛠️ Testing Skill Analysis...")

try:
    # Test basic skill frequency
    top_skills = skill_analyzer.analyze_skill_frequency(skills_df, 10)
    print(f"✅ Skill frequency: {len(top_skills)} skills analyzed")
    print("Top 5 skills:")
    for i, row in top_skills.head().iterrows():
        print(f"  {i+1}. {row['skill']} ({row['prevalence']:.1f}%)")
    
    # Test skill categories
    categories = skill_analyzer.analyze_skill_categories(skills_df, 5)
    print(f"✅ Skill categories: {len(categories)} categories analyzed")
    
except Exception as e:
    print(f"❌ Skill analysis failed: {e}")
```

### 4.3 Test Role Analysis


```python
print("\n🎭 Testing Role Analysis...")

try:
    # Test role-based skills
    role_skills = role_analyzer.analyze_skills_by_role(skills_df, 5)
    print(f"✅ Role analysis: {len(role_skills)} roles analyzed")
    
    for role, skills_data in list(role_skills.items())[:3]:
        top_skill = skills_data.iloc[0]['skill'] if not skills_data.empty else "No skills"
        print(f"  - {role}: {top_skill}")
    
    # Test seniority analysis
    seniority_skills = role_analyzer.analyze_skills_by_seniority(skills_df, 5)
    print(f"✅ Seniority analysis: {len(seniority_skills)} levels analyzed")
    
except Exception as e:
    print(f"❌ Role analysis failed: {e}")
```

### 4.4 Test Trend Analysis

```python
print("\n📈 Testing Trend Analysis...")

try:
    # Test trend analysis with relaxed parameters
    trends = trend_analyzer.analyze_skill_trends_enhanced(
        skills_df,
        min_prevalence=0.5,  # Lower threshold for testing
        min_avg_mentions=2,  # Lower threshold for testing
        min_months=3         # Lower threshold for testing
    )
    
    if not trends.empty:
        print(f"✅ Trend analysis: {len(trends)} trends found")
        print("Trend categories distribution:")
        print(trends['trend_category'].value_counts())
    else:
        print("⚠️ No trends found - may need more temporal data")
        
except Exception as e:
    print(f"❌ Trend analysis failed: {e}")
```

## 5. Test Filter Functionality
### 5.1 Test Data Filtering

```python
print("\n🎛️ Testing Data Filtering...")

try:
    # Test filtering on actual data
    test_filters = [
        {},
        {'country': 'USA'},
        {'standardized_title': 'Data Scientist'},
        {'seniority_level': 'Senior'}
    ]
    
    for i, filters in enumerate(test_filters, 1):
        print(f"\nTest {i}: {filters}")
        
        # Apply filters manually
        filtered_main = df.copy()
        filtered_skills = skills_df.copy()
        
        # Apply country filter
        if 'country' in filters and filters['country'] in df['country'].unique():
            country = filters['country']
            filtered_main = filtered_main[filtered_main['country'] == country]
            filtered_skills = filtered_skills[filtered_skills['country'] == country]
            print(f"  Applied country filter: {country}")
        
        # Apply role filter  
        if 'standardized_title' in filters and filters['standardized_title'] in df['standardized_title'].unique():
            role = filters['standardized_title']
            filtered_main = filtered_main[filtered_main['standardized_title'] == role]
            filtered_skills = filtered_skills[filtered_skills['standardized_title'] == role]
            print(f"  Applied role filter: {role}")
            
        # Apply seniority filter
        if 'seniority_level' in filters and filters['seniority_level'] in df['seniority_level'].unique():
            seniority = filters['seniority_level']
            filtered_main = filtered_main[filtered_main['seniority_level'] == seniority]
            filtered_skills = filtered_skills[filtered_skills['seniority_level'] == seniority]
            print(f"  Applied seniority filter: {seniority}")
        
        print(f"  Results: {len(filtered_main)} main records, {len(filtered_skills)} skill records")
        
        # Test analysis on filtered data
        if not filtered_skills.empty:
            filtered_top_skills = skill_analyzer.analyze_skill_frequency(filtered_skills, 3)
            print(f"  Top filtered skills: {list(filtered_top_skills['skill'])}")
            
except Exception as e:
    print(f"❌ Filter testing failed: {e}")
```

## 6. Test Advanced Analysis Features
### 6.1 Test Skill Combinations

```python
print("\n🔗 Testing Skill Combinations...")

try:
    combinations = skill_analyzer.analyze_skill_combination_prevalence(skills_df, 5)
    if not combinations.empty:
        print(f"✅ Skill combinations: {len(combinations)} pairs analyzed")
        for i, row in combinations.head().iterrows():
            print(f"  {i+1}. {row['skill_1']} + {row['skill_2']} ({row['mentions']} mentions)")
    else:
        print("⚠️ No skill combinations found")
        
except Exception as e:
    print(f"❌ Skill combinations failed: {e}")
```

### 6.2 Test Seniority Patterns

```python
print("\n👨‍💼 Testing Seniority Patterns...")

try:
    # Get progression data first
    top_skills_list = skill_analyzer.analyze_skill_frequency(skills_df, 10)['skill'].tolist()
    progression_data = role_analyzer.analyze_skill_progression_data(skills_df, top_skills_list[:5])
    
    # Test seniority patterns
    seniority_patterns = seniority_analyzer.analyze_seniority_specific_patterns(skills_df, progression_data)
    print(f"✅ Seniority patterns: {len(seniority_patterns)} levels analyzed")
    
    for level, patterns in seniority_patterns.items():
        print(f"  - {level}: {patterns.get('total_jobs', 0)} jobs, {len(patterns.get('top_skills', []))} top skills")
        
except Exception as e:
    print(f"❌ Seniority patterns failed: {e}")
```
